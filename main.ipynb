```python
import io
import pandas as pd

def read_clean_csv(path):
    # The CSVs have stray quotes wrapping each line, so pandas sees one column; strip them first
    with open(path, encoding="utf-8") as f:
        text = f.read().replace('"', '')
    return pd.read_csv(io.StringIO(text))

df = read_clean_csv("Data/investments_maturity.csv")
print(df.head())
totals = df.groupby("bank")["amount_crore"].sum()
print(totals)
```
